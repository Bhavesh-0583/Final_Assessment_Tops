This notebook covers basic perceptron operations, the XOR limitation, and simple multilayer neural network examples using NumPy.

## 1. Perceptron for Flipkart Review Classification

```python
import numpy as np

# Features: [review score, number of helpful votes]
review_data = np.array([
    [85, 6],
    [70, 5],
    [30, 2],
    [25, 1],
    [90, 5],
    [15, 0]
], dtype=float)

# 1 = positive review, 0 = negative review
labels = np.array([1, 1, 0, 0, 1, 0])

# Normalize each feature column
feature_max = review_data.max(axis=0)
review_data = review_data / feature_max

# Initialize perceptron parameters
w = np.zeros(review_data.shape[1])
b = 0.0

lr = 0.1
iterations = 20

def activate(value):
    return int(value >= 0)

# Train the perceptron
for _ in range(iterations):
    for features, target in zip(review_data, labels):
        score = np.dot(features, w) + b
        predicted_label = activate(score)
        difference = target - predicted_label

        w = w + lr * difference * features
        b = b + lr * difference

print("Learned weights:", w)
print("Learned bias:", b)

print("\nModel predictions")
for features, actual_label in zip(review_data, labels):
    prediction = activate(np.dot(features, w) + b)
    print(f"Features: {features} | Predicted: {prediction} | Actual: {actual_label}")
```

## 2. Weighted Sum and Step Activation

```python
import numpy as np

def calculate_perceptron(features, parameters, intercept):
    total_value = np.dot(features, parameters) + intercept
    predicted_class = int(total_value >= 0)
    return total_value, predicted_class

parameters = np.array([0.6, 0.8])

examples = [
    {"features": np.array([1, 1]), "bias": -1.0},
    {"features": np.array([0, 1]), "bias": -0.2},
    {"features": np.array([1, 0]), "bias": -0.5},
    {"features": np.array([-0.3, 0.4]), "bias": -0.4}
]

for number, example in enumerate(examples, start=1):
    value, prediction = calculate_perceptron(
        example["features"],
        parameters,
        example["bias"]
    )

    print(f"Case {number}")
    print("Input values :", example["features"])
    print("Weighted sum :", round(float(value), 3))
    print("Step output  :", prediction)
    print("-" * 32)
```

## 3. Why a Single-Layer Perceptron Cannot Learn XOR

```python
import numpy as np

# XOR input-output pairs
xor_inputs = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

xor_targets = np.array([0, 1, 1, 0])

xor_weights = np.zeros(2)
xor_bias = 0.0

rate = 0.1
training_rounds = 20

def binary_step(value):
    return int(value >= 0)

# Try training a single-layer perceptron
for _ in range(training_rounds):
    for sample, target in zip(xor_inputs, xor_targets):
        predicted = binary_step(np.dot(sample, xor_weights) + xor_bias)
        error_value = target - predicted

        xor_weights += rate * error_value * sample
        xor_bias += rate * error_value

print("Final XOR weights:", xor_weights)
print("Final XOR bias:", xor_bias)

print("\nResults after training:")
for sample, target in zip(xor_inputs, xor_targets):
    predicted = binary_step(np.dot(sample, xor_weights) + xor_bias)
    print(f"{sample} -> Predicted: {predicted}, Actual: {target}")

print("\nObservation: XOR is not linearly separable, so a single-layer perceptron cannot solve it perfectly.")
```

## 4. Multilayer Perceptron (MLP) for XOR Using NumPy

```python
import numpy as np

xor_data = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

def sigmoid_function(value):
    return 1 / (1 + np.exp(-value))

# Hidden layer parameters
hidden_weights = np.array([
    [20, 20],
    [20, 20]
])

hidden_bias = np.array([-10, -30])

# Output layer parameters
output_weights = np.array([
    [20],
    [-20]
])

output_bias = np.array([-10])

print("XOR using a multilayer perceptron\n")

for sample in xor_data:
    # Forward pass through hidden layer
    hidden_sum = sample @ hidden_weights + hidden_bias
    hidden_layer = sigmoid_function(hidden_sum)

    # Forward pass through output layer
    output_sum = hidden_layer @ output_weights + output_bias
    probability = sigmoid_function(output_sum)

    predicted_class = int(probability.item() >= 0.5)

    print(
        f"Input: {sample} | "
        f"Probability: {probability.item():.4f} | "
        f"Prediction: {predicted_class}"
    )
```

## 5. MLP for Zomato Trending Restaurant Classification

```python
import numpy as np

# Features: [restaurant rating, number of orders/reviews]
restaurant_data = np.array([
    [4.8, 250],
    [4.6, 180],
    [3.5, 40],
    [3.8, 60],
    [4.9, 300]
], dtype=float)

# Scale features separately
restaurant_data[:, 0] /= 5.0
restaurant_data[:, 1] /= 300.0

def logistic_activation(value):
    return 1 / (1 + np.exp(-value))

# First layer
layer1_weights = np.array([
    [3.0, -2.0],
    [2.5, 2.5]
])

layer1_bias = np.array([-1.5, -1.0])

# Final layer
layer2_weights = np.array([
    [2.5],
    [3.5]
])

layer2_bias = np.array([-2.0])

print("Restaurant Trending Prediction\n")

for index, restaurant in enumerate(restaurant_data, start=1):
    hidden_values = logistic_activation(
        restaurant @ layer1_weights + layer1_bias
    )

    final_probability = logistic_activation(
        hidden_values @ layer2_weights + layer2_bias
    )

    is_trending = int(final_probability.item() >= 0.5)

    print(f"Restaurant {index}")
    print(f"Input features : {restaurant}")
    print(f"Probability    : {final_probability.item():.4f}")
    print(f"Trending class : {is_trending}")
    print("-" * 35)
```
